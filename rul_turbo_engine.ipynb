```python
columns = ['Unit_Number', 'Time_in_cycles', 'Setting_1', 'Setting_2', 'TRA', 'T2', 'T24', 'T30', 'T50', 'P2', 'P15', 'P30', 'Nf', 'Nc', 'epr', 'Ps30', 'Phi', 'NRf', 'NRc', 'BPR', 'farB', 'htBleed', 'Nf_dmd', 'PCNfr_dmd', 'W31', 'W32']
fd_001_train.columns = columns
fd_001_test.columns = columns
```

```python
# Plotting Histogram for each columns
for col in fd_001_train.columns:
    plt.figure(figsize=(6,4))
    sns.histplot(fd_001_train[col], bins = 30, kde = True)
    plt.title(f'Histplot of {col}')
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.show()
```

```python
# Delete columns with constant values that do no carry information about the state of the unit
fd_001_train = fd_001_train.drop(columns = ['TRA', 'T2', 'P2', 'P15', 'epr', 'farB', 'Nf_dmd', 'PCNfr_dmd'])
```

```python
fd_001_train['Life_Ratio'] = fd_001_train['Time_in_cycles']/fd_001_train.groupby('Unit_Number')['Time_in_cycles'].transform('max')
```

```python
sns.heatmap(fd_001_train.corr(), annot = True, cmap = 'RdYlGn', linewidth = 0.2)
fig = plt.gcf()
fig.set_size_inches(20,20)
plt.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error
```

```python
X = fd_001_train.drop(columns = ['Unit_Number', 'Setting_1', 'Setting_2', 'Life_Ratio'])
Y = fd_001_train['Life_Ratio']
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size = 0.2, random_state = 42)
```

```python
# Train Random Forest Regressor
model = RandomForestRegressor(n_estimators = 100, random_state = 42)
model.fit(X_train, Y_train)
```

```python
# Predict and Evaluate
Y_pred = model.predict(X_test)
mse = mean_squared_error(Y_test, Y_pred)
print("Mean Square Error: ", mse)
```

```python
# Plot the predicted vs True life ratio for a specific unit number
unit_id = fd_001_train['Unit_Number'].unique()[10]
unit_data = fd_001_train[fd_001_train['Unit_Number'] == unit_id]
true_life_ratio = unit_data['Life_Ratio'].values
pred_life_ratio = model.predict(unit_data.drop(columns = ['Unit_Number', 'Setting_1', 'Setting_2', 'Life_Ratio']))

plt.figure(figsize=(8,5))

plt.plot(
    unit_data['Time_in_cycles'],
    true_life_ratio,
    label="True Life Ratio",
    marker="o"
)

plt.plot(
    unit_data['Time_in_cycles'],
    pred_life_ratio,
    label="Predicted Life Ratio",
    marker="*"
)

plt.xlabel("Time in Cycles")
plt.ylabel("Life Ratio")
plt.legend()
plt.grid(True)
plt.show()




                    
```

```python
train_data_processed = []

for idx in range(1, 5):
    train_data = pd.read_csv(
        rf"D:\My Research\RUL of Turbo Engine\archive\CMaps\test_FD00{idx}.txt",
        sep=" ",
        header=None
    )

    # Drop last two empty columns (26, 27)
    train_data = train_data.drop(columns=[26, 27])

    # Assign proper column names
    columns = [
        'Unit_Number', 'Time_in_cycles',
        'Setting_1', 'Setting_2', 'TRA',
        'T2', 'T24', 'T30', 'T50',
        'P2', 'P15', 'P30',
        'Nf', 'Nc', 'epr',
        'Ps30', 'Phi', 'NRf',
        'NRc', 'BPR', 'farB',
        'htBleed', 'Nf_dmd',
        'PCNfr_dmd', 'W31', 'W32'
    ]
    train_data.columns = columns

    # Life Ratio
    train_data['Life_Ratio'] = (
        train_data['Time_in_cycles'] /
        train_data.groupby('Unit_Number')['Time_in_cycles'].transform('max')
    )

    # Drop unused sensors
    train_data = train_data.drop(
        columns=['TRA', 'T2', 'P2', 'P15', 'epr', 'farB', 'Nf_dmd', 'PCNfr_dmd']
    )

    train_data_processed.append(train_data)

    print(f"Processed test_FD00{idx}.txt, rows: {train_data.shape[0]}")

# Combine all datasets
train_data_processed = pd.concat(train_data_processed, ignore_index=True)
```
